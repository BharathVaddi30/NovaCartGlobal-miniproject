```sql
CREATE OR REPLACE TABLE novacart_catalog.003_gold.sales_datacube AS
SELECT 
  d.full_date,
  d.year,
  d.quarter,
  d.month_name,
  d.year_month,
  
  p.product_id,
  p.product_name,
  
  c.customer_id,
  c.customer_name,
  c.country_code as customer_country,
  
  f.order_key as order_id,
  f.Channel as channel,
  f.Country_code as order_country_code,
  f.Order_status as order_status,
  
  SUM(f.Quantity) as total_quantity,
  SUM(f.line_total_original) as total_revenue_original_currency,
  SUM(f.line_total_original * f.Exchange_rate_to_usd) as total_revenue_usd,
  COUNT(DISTINCT f.order_key) as order_count,
  COUNT(f.order_item_key) as order_item_count,
  AVG(f.unit_price_original) as avg_unit_price
  
FROM novacart_catalog.003_gold.gold_facts_table f

LEFT JOIN novacart_catalog.003_gold.dim_date d
  ON DATE(f.Order_date) = d.full_date

LEFT JOIN novacart_catalog.003_gold.dim_products p
  ON f.product_key = p.product_id

LEFT JOIN novacart_catalog.003_gold.dim_customers c
  ON f.customer_key = c.customer_id

GROUP BY 
  d.full_date,
  d.year,
  d.quarter,
  d.month_name,
  d.year_month,
  p.product_id,
  p.product_name,
  c.customer_id,
  c.customer_name,
  c.country_code,
  f.order_key,
  f.Channel,
  f.Country_code,
  f.Order_status
```
