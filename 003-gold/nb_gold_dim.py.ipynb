###  Dim_date

```sql
CREATE OR REPLACE TABLE novacart_catalog.003_gold.dim_date
AS
WITH date_spine AS (
  SELECT EXPLODE(SEQUENCE(
    DATE '2023-01-01',
    DATE '2025-12-31',
    INTERVAL 1 DAY
  )) AS date_value
)
SELECT 
  CAST(DATE_FORMAT(date_value, 'yyyyMMdd') AS INT) AS date_key,
  date_value AS full_date,
  YEAR(date_value) AS year,
  QUARTER(date_value) AS quarter,
  MONTH(date_value) AS month_number,
  DATE_FORMAT(date_value, 'MMMM') AS month_name,
  DATE_FORMAT(date_value, 'yyyy-MM') AS year_month,
  DAYOFWEEK(date_value) AS day_of_week,
  DATE_FORMAT(date_value, 'EEEE') AS day_name
FROM date_spine;
```

```sql
select * from novacart_catalog.003_gold.dim_date
```
