### Fact Table

```sql
CREATE OR REPLACE TABLE novacart_catalog.003_gold.gold_facts_table
AS
SELECT 
  oi.Order_item_id AS order_item_key,
  oi.Order_id AS order_key,
  oi.Product_id AS product_key,
  o.Customer_id AS customer_key,
  o.Order_date,
  o.Order_status,
  o.Channel,
  o.Country_code,
  oi.Quantity,
  oi.Unit_price AS unit_price_original,
  oi.Line_total AS line_total_original,
  o.Currency AS original_currency,
  COALESCE(er.Exchange_rate_to_usd, 1.0) AS Exchange_rate_to_usd
FROM novacart_catalog.002_silver.order_items oi
INNER JOIN novacart_catalog.002_silver.orders o
  ON oi.Order_id = o.Order_id
LEFT JOIN novacart_catalog.002_silver.exchange_rates er
  ON o.Currency = er.Currency_code
WHERE o.Order_status = 'completed';
```

```sql

SELECT 
  order_item_key,
  order_key,
  customer_key,
  product_key,
  order_date,
  channel,
  country_code,
  quantity,
  Exchange_rate_to_usd
  order_status
FROM novacart_catalog.003_gold.gold_facts_table
ORDER BY order_date DESC
LIMIT 10;
```
